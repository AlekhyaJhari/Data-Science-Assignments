```python
from sklearn.preprocessing import StandardScaler

numerical_cols = ['age', 'fnlwgt', 'education_num', 'capital_gain', 'capital_loss', 'hours_per_week']

scaler = StandardScaler()

df_standard = df.copy()
df_standard[numerical_cols] = scaler.fit_transform(df_standard[numerical_cols])

df_standard.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()

df_minmax = df.copy()
df_minmax[numerical_cols] = minmax.fit_transform(df_minmax[numerical_cols])

df_minmax.head()
```

Explain the difference

Standard Scaling: Transforms data to have a mean of 0 and a standard deviation of 1. It is preferred for algorithms such as Logistic Regression, SVM, and PCA.

Min-Max Scaling: Scales values to the range 0–1. It is useful for Neural Networks and distance-based algorithms like KNN.

```python
# Find categorical columns with less than 5 unique values
small_cat_cols = [col for col in df.select_dtypes(include='object').columns
                  if df[col].nunique() < 5]

print(small_cat_cols)

df_onehot = pd.get_dummies(df, columns=small_cat_cols)

df_onehot.head()
```

```python
from sklearn.preprocessing import LabelEncoder

df_label = df.copy()

le = LabelEncoder()

for col in df_label.select_dtypes(include='object').columns:
    df_label[col] = le.fit_transform(df_label[col])

df_label.head()
```

One-Hot Encoding:
• Creates a new binary column for each category.
• Best for columns with fewer categories.

Label Encoding:
• Assigns a numeric value to each category.
• Best for columns with many categories.

```python
df['capital_total'] = df['capital_gain'] - df['capital_loss']

df['age_per_education'] = df['age'] / df['education_num']

df.head()
```

Feature 1: capital_total

Combines capital gain and capital loss into one feature to represent the overall capital amount.

Feature 2: age_per_education

Represents the relationship between a person’s age and education level, which may help improve prediction.

```python
import numpy as np

df['fnlwgt_log'] = np.log1p(df['fnlwgt'])

df[['fnlwgt', 'fnlwgt_log']].head()
```

Log Transformation

Log transformation reduces skewness in numerical data and makes the distribution more normal, which can improve machine learning model performance.
