# Assignment 6: Estimation and Confidence Intervals

## Objective
To construct 99% confidence intervals for the mean durability of print-heads using:
- Sample Standard Deviation (t-distribution)
- Known Population Standard Deviation (z-distribution)

```python
import numpy as np
from scipy import stats
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25,
        1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
std = np.std(data, ddof=1)

print("Sample Size:", n)
print("Sample Mean:", mean)
print("Sample Standard Deviation:", std)
```

```python
confidence = 0.99
t_value = stats.t.ppf((1 + confidence) / 2, df=n-1)

margin_error = t_value * (std / np.sqrt(n))

lower = mean - margin_error
upper = mean + margin_error

print("99% Confidence Interval (t-distribution):")
print((lower, upper))
```

```python
population_std = 0.2
z_value = stats.norm.ppf(0.995)
margin_error = z_value * (population_std / np.sqrt(n))
lower = mean - margin_error
upper = mean + margin_error
print("99% Confidence Interval (known Population SD):")
print((lower, upper))
```

# Conclusion

- Calculated the sample mean and sample standard deviation.
- Constructed a 99% confidence interval using the t-distribution because the population standard deviation was unknown.
- Constructed another 99% confidence interval using the z-distribution assuming the population standard deviation was known.
- The confidence intervals provide an estimated range within which the true population mean durability is likely to lie with 99% confidence.
