```python
patient_df = pd.read_csv("Patient_Data.csv")
billing_df = pd.read_csv("Billing_Data.csv")
```

# Task-2

```python
billing_columns = patient_df[['PatientID', 'Department', 'Doctor', 'BillAmount']]
billing_columns
```

# Task-3

```python
patient_df = patient_df.drop(columns=['ReceptionistID', 'CheckInTime'])
patient_df.head()
```

# Task-4

```python
patient_df.groupby('Department')['BillAmount'].sum()
```

# Task-5

```python
patient_df = patient_df.drop_duplicates(subset='PatientID')
```

# Task-6

```python
patient_df['BillAmount'] = patient_df['BillAmount'].fillna(patient_df['BillAmount'].mean())
patient_df
```


# Task-7

```python
merged_df = pd.merge(patient_df, billing_df, on='PatientID')
merged_df
```

# Task-8

```python
new_patient = pd.DataFrame({
    'PatientID':[106],
    'Name':['John'],
    'Department':['Orthopedics'],
    'Doctor':['Dr. Alex'],
    'BillAmount':[6500]
})

patient_df = pd.concat([patient_df, new_patient], ignore_index=True)
patient_df
```

# Task-9

```python
new_columns = billing_df[['InsuranceCovered','FinalAmount']]

final_df = pd.concat([patient_df, new_columns], axis=1)
final_df
```
