```python
# Question 1

#create a list
numbers = [10, 20, 30, 40, 50]
print("Original List:", numbers)

# append()
numbers.append(60)
print("Original List:", numbers)

#extend()
numbers.extend([70, 80])
print("After exetend:", numbers)

# insert()
numbers.insert(2,27)
print("After insert:", numbers)

# remove()
numbers.remove(40)
print("After remove:", numbers)

#pop()
numbers.pop()
print("After p0p:", numbers)

# index()
print("Index of 30:", numbers.index(30))

# sort()
numbers.sort()
print("After sort:", numbers)

# reverse()
numbers.reverse()
print("After reverse:", numbers)

# clear()
numbers.clear()
print("After clear:", numbers)
```

```python
# Question-2
students = ("Rohan", "Alekhya", "Sherli")
print("Original Tuple:", students)
students[1] = "Sujatha"
print(students)
```

Tuple is immutable in python.
It does not allow changing, adding, or removing elements after it is created.

```python
# Question 7
number = 8
if number > 0:
    print("The number is positive.")
    if number % 2 == 0:
        print("The number is also even.")
    else:
        print("The number is odd.")
elif number == 0:
    print("The number is zero.")
else:
    print("The number is negative.")
```
